This notebook runs CS, CCS and LS-CCS on the Michalewicz benchmark.

```python
import sys
from pathlib import Path
ROOT = Path.cwd()
if (ROOT / 'src').exists():
    sys.path.insert(0, str(ROOT / 'src'))
else:
    sys.path.insert(0, str(ROOT.parent / 'src'))

from benchmark_functions import michalewicz
from cuckoo_search import cuckoo_search
from chaotic_cuckoo_search import chaotic_cuckoo_search
from logistic_sine_ccs import ls_ccs
```

```python
configs = {
    'CS': cuckoo_search,
    'CCS': chaotic_cuckoo_search,
    'LS-CCS': ls_ccs,
}

for name, algorithm in configs.items():
    best, fitness, convergence = algorithm(michalewicz, dim=2, lb=0, ub=3.141592653589793)
    print(name, 'best fitness =', fitness, 'best position =', best)
```

```python
import matplotlib.pyplot as plt

for name, algorithm in configs.items():
    _, _, convergence = algorithm(michalewicz, dim=2, lb=0, ub=3.141592653589793)
    plt.plot(convergence, label=name)

plt.xlabel('Iteration')
plt.ylabel('Best fitness')
plt.title('Cuckoo Search convergence — Michalewicz (2D)')
plt.grid(True)
plt.legend()
plt.show()
```
